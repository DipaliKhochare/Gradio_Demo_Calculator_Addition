```python
import gradio as gr

def add_numbers(Num1, Num2):
    return Num1 + Num2

# Define the interface
demo = gr.Interface(
    fn=add_numbers,
    inputs=[gr.Number(), gr.Number()], # Create two numerical input fields where users can enter numbers
    outputs=gr.Number() # Create numerical output fields
)

# Close any existing active Gradio interfaces to free the port
gr.close_all()

# Launch the interface on any free port
demo.launch(server_name="127.0.0.1", share=True)
```
